**PERTANYAAN PRKTIKUM D3**

1. Without Matplotlib. The image is displayed cleanly as is. The colors look normal because OpenCV reads images in the BGR color format by default.
With Matplotlib. The image is displayed inside a plot, complete with coordinate axes that act like a "ruler" showing the pixel dimensions on the edges.

2. They differ in their limit sizes and memory consumption. int32 takes up more capacity and memory than int16.

3. The cv2_imshow command is a special workaround for Google Colab. Regular OpenCV functions try to open images in a new pop-up window, which does not work in web browsers

4. This acts as a shortcut so the Python program can fetch and read an image directly from an internet URL link.
